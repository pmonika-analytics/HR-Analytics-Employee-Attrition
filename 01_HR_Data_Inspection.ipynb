```python
# Attrition Distribution
plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x='Attrition'
)

plt.title('Employee Attrition Distribution')
plt.xlabel('Attrition Status')
plt.ylabel('Number of Employees')

plt.show()
```

```python
attrition = df['Attrition'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    attrition,
    labels=attrition.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Employee Attrition Percentage')

plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.countplot(
    data=df,
    x='Department',
    hue='Attrition'
)

plt.title('Department-wise Employee Attrition')
plt.xlabel('Department')
plt.ylabel('Employee Count')

plt.show()
```

```python
department_attrition = pd.crosstab(
    df['Department'],
    df['Attrition']
)

department_attrition
```

```python
department_attrition_rate = (
    df.groupby('Department')['Attrition']
    .apply(lambda x: (x == 'Yes').mean() * 100)
)

department_attrition_rate
```

```python
# Analysis A: Overtime vs Attrition

df.groupby('OverTime')['Attrition'].value_counts()
```

```python
df.groupby('OverTime')['Attrition'].apply(
    lambda x: (x == 'Yes').mean() * 100
).round(2)
```

```python
# Analysis B: Age Group vs Attrition
df['AgeGroup'] = pd.cut(
    df['Age'],
    bins=[17, 25, 35, 45, 55, 65],
    labels=['18-25', '26-35', '36-45', '46-55', '56-65']
)
df.groupby('AgeGroup', observed=True)['Attrition'].apply(
    lambda x: (x == 'Yes').mean() * 100
).round(2)
```

```python
# Analysis C: Job Satisfaction vs Attrition
df.groupby('JobSatisfaction')['Attrition'].apply(
    lambda x: (x == 'Yes').mean() * 100
).round(2)
```

```python
# Analysis D: Salary vs Attrition
df.groupby('Attrition')['MonthlyIncome'].mean().round(2)
```

```python
sns.countplot(data=df, x='OverTime', hue='Attrition')
plt.title('Overtime vs Employee Attrition')
plt.show()
```

```python
sns.countplot(data=df, x='AgeGroup', hue='Attrition')
plt.title('Age Group vs Employee Attrition')
plt.show()
```
